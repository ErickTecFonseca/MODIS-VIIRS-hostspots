```python
def process_modis(path, country):
    df = pd.read_csv(path)

    # Normalice columns names
    df.columns = df.columns.str.strip().str.lower()

    # Create country column
    df["country"] = country

    # Transform to date time
    df["acq_date"] = pd.to_datetime(
        df["acq_date"],
        errors="coerce"
    )

    # Source file
    df["source_file"] = path.name

    # Numeric Cols
    numeric_columns = [
        "latitude",
        "longitude",
        "brightness",
        "scan",
        "track",
        "bright_t31",
        "frp"
    ]

    # Convert to numeric columns
    for column in numeric_columns:
        df[column] = pd.to_numeric(
            df[column],
            errors="coerce"
        )

    return df

path = Path(r"data\raw\MODIS\modis_2000_all_countries\modis\2000\modis_2000_Afghanistan.csv")
proces_modis(path, "a")
```

```python
import pandas as pd

def process_viirs(path, country):

    df = pd.read_csv(path)

    df.columns = df.columns.str.strip().str.lower()

    # Contry col
    df["country"] = country

    # Numeric cols
    numeric_columns = [
        "latitude",
        "longitude",
        "bright_ti4",
        "bright_ti5",
        "scan",
        "track",
        "frp"
    ]

    df["source_file"] = path.name

    for column in numeric_columns:
        df[column] = pd.to_numeric(
            df[column],
            errors="coerce"
        )

    # Date
    df["acq_date"] = pd.to_datetime(
        df["acq_date"],
        errors="coerce"
    )

    return df

path = Path(r"data\raw\VIIRS\viirs-jpss1_2018_all_countries\viirs-jpss1\2018\viirs-jpss1_2018_Afghanistan.csv")
process_viirs(path, "a")
```

```python
years = np.arange(2000, 2026, 1)
raw_path = Path("data/raw")

# Obtener todos los CSV
files = list(raw_path.rglob("*.csv"))

# Agrupar archivos por año
files_by_year = {}

for file in files:
    fileName = file.name

    # Find year
    for y in years:
        if str(y) in fileName:
            year = int(y)

    if "viirs" in fileName:
        country = fileName.replace(f"viirs-jpss1_{year}_", "")
        country = country.replace(".csv", "")

    elif "modis" in fileName:
        country = fileName.replace(f"modis_{year}_", "")
        country = country.replace(".csv", "")

    else:
        pass

    files_by_year.setdefault(year, []).append(
        (file, country)
    )

# Process each year
for year, year_files in files_by_year.items():
    viirsDataFrames = []
    modisDataFrames = []

    for file, country in year_files:
        fileName = file.name

        if "viirs" in fileName:
            df = process_viirs(file, country)
            viirsDataFrames.append(df)
        
        elif "modis" in fileName:
            df = process_modis(file, country)
            modisDataFrames.append(df)

    # Combine all contries
    if len(viirsDataFrames) > 0: # Only created from 2018 to 2025
        final_viirs_df = pd.concat(
            viirsDataFrames,
            ignore_index=True
        )

    final_modis_df = pd.concat(
            modisDataFrames,
            ignore_index=True
    )

    # Create a viirs dir
    viirs_processed_path = Path(r"data\processed\VIIRS")
    viirs_output_dir = viirs_processed_path / f"year={year}"
    viirs_output_dir.mkdir(parents=True, exist_ok=True)

    # Create a modis dir
    modis_processed_path = Path(r"data\processed\MODIS")
    modis_output_dir = modis_processed_path / f"year={year}"
    modis_output_dir.mkdir(parents=True, exist_ok=True)

    # Save Parquet
    if len(viirsDataFrames) > 0: # Only save from 2018 to 2025
        final_viirs_df.to_parquet(
            viirs_output_dir / "data.parquet",
            index=False
        )

    final_modis_df.to_parquet(
        modis_output_dir / "data.parquet",
        index=False
    )

    print(f"Año {year}, Viirs: {len(final_viirs_df)} detecciones")
    print(f"Año {year}, Modis:: {len(final_modis_df)} detecciones")
```

```python
# By year and country
years = np.arange(2000, 2026, 1)
raw_path = Path("data/raw")

# Obtener todos los CSV
files = list(raw_path.rglob("*.csv"))

# Agrupar archivos por año
files_by_year = {}

for file in files:
    fileName = file.name

    # Find year
    for y in years:
        if str(y) in fileName:
            year = int(y)

    if "viirs" in fileName:
        country = fileName.replace(f"viirs-jpss1_{year}_", "")
        country = country.replace(".csv", "")

    elif "modis" in fileName:
        country = fileName.replace(f"modis_{year}_", "")
        country = country.replace(".csv", "")

    else:
        pass

    files_by_year.setdefault(year, []).append(
        (file, country)
    )

# Process each year
for year, year_files in files_by_year.items():

    files_by_country_viirs = {}
    files_by_country_modis = {}

    for file, country in year_files:
        fileName = file.name

        if "viirs" in fileName:
            df = process_viirs(file, country)
            files_by_country_viirs.setdefault(country, []).append(
                df
            )
        
        elif "modis" in fileName:
            df = process_modis(file, country)
            files_by_country_modis.setdefault(country, []).append(
                df
            )


    # Save per contries
    if len(files_by_country_viirs) > 0: # Only created from 2018 to 2025
        for country, df in files_by_country_viirs.items():
            final_viirs_df = pd.concat(
                df,
                ignore_index=True
            )

            # Create a viirs dir
            viirs_processed_path = Path(r"data\processed_country\VIIRS")
            viirs_output_dir = viirs_processed_path / f"year={year}" / f"country={country}"
            viirs_output_dir.mkdir(parents=True, exist_ok=True)

            # Save parquet
            final_viirs_df.to_parquet(
                    viirs_output_dir / "data.parquet",
                    index=False
                )

    for country, df in files_by_country_modis.items():
        final_modis_df = pd.concat(
                df,
                ignore_index=True
        )

        # Create a modis dir
        modis_processed_path = Path(r"data\processed_country\MODIS")
        modis_output_dir = modis_processed_path / f"year={year}" / f"country={country}"
        modis_output_dir.mkdir(parents=True, exist_ok=True)

        # Save Parquet
        final_modis_df.to_parquet(
            modis_output_dir / "data.parquet",
            index=False
        )

        print(f"Año {year}, País {country}, Viirs: {len(final_viirs_df)} detecciones")
        print(f"Año {year}, País {country}, Modis: {len(final_modis_df)} detecciones")
```
